*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution here
def winner_verification(game_board):
    for row in game_board:
        if all(value == 1 for value in row):
            return 1
        elif all (value == 2 for value in row):
            return 2
        #elif any (value == 0 for value)
    for col in range(len(game_board)):
        if all(game_board[row][col] == 1 for row in range(len(game_board))):
            return 1
        elif all(game_board[row][col] == 2 for  row in range(len(game_board))):
            return 2
    if all (game_board[i][i] == 1 for i in range(len(game_board))):
        return 1
    if all (game_board[i][i] == 2 for i in range(len(game_board))):
            return 2
    if all(game_board[i][len(game_board)-1-i] == 1 for i in range(len(game_board))):
        return 1
    elif all(game_board[i][len(game_board)-1-i] == 2 for i in range(len(game_board))):
        return 2
    if any(0 in row for row in game_board):
        return -1
    return 0
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
# Write your solution here
def place_marker(game_board, player, row, col):
    if game_board[row][col] == 0:
        game_board[row][col] = player
        return True
    return False
    
````

*Exercise 8:* Write a function that is called with a board and player number, takes input from the player using Python's `input()`, and modifies the board using your function from Exercise 7.

Keep asking for input until the player enters a valid location that results in a valid move.


````python
# Write your solution here
def player_turn(game_board, player):
    while True:
        location = input("Enter Location (or ESC to quit): ")

        if location.upper() == "ESC":
            return False
        
        if len(location) < 2:
            print("Invalid location. Try again.")
            continue

        try:
            if make_move(game_board, player, location):
                break
            else:
                print("Location already taken. Try again.")
        except:
            print("Invalid location. Try again.")
````

````python
# Test your solution here
game_board = [
    [0, 0, 0],
    [0, 0, 0],
    [0, 0, 0]
]

player_turn(game_board, 1,)
print(game_board)
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Write your solution here
def computer_move(game_board):
    # Try to win first
    for row in range(len(game_board)):
        for col in range(len(game_board)):
            if game_board[row][col] == 0:
                game_board[row][col] = 2

                if winner_verification(game_board) == 2:
                    return

                game_board[row][col] = 0

    # Block Player 1 from winning
    for row in range(len(game_board)):
        for col in range(len(game_board)):
            if game_board[row][col] == 0:
                game_board[row][col] = 1

                if winner_verification(game_board) == 1:
                    game_board[row][col] = 2
                    return

                game_board[row][col] = 0

    # Otherwise choose the first empty spot
    for row in range(len(game_board)):
        for col in range(len(game_board)):
            if game_board[row][col] == 0:
                game_board[row][col] = 2
                return


def computer_game():
    game_board = [[0] * 3 for i in range(3)]

    while winner_verification(game_board) == -1:
        board_game_creation(game_board)
        print("Your turn")

        if player_turn(game_board, 1) == False:
            print("Game cancelled")
            return

        if winner_verification(game_board) != -1:
            break

        print("Computer's turn")
        computer_move(game_board)

    board_game_creation(game_board)

    winner = winner_verification(game_board)

    if winner == 0:
        print("Draw")
    elif winner == 1:
        print("You win!")
    else:
        print("Computer wins!")
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
# Write your solution here
def minimax(game_board, depth, max_depth, maximizing):
    result = winner_verification(game_board)

    if result == 2:
        return 10 - depth
    elif result == 1:
        return depth - 10
    elif result == 0:
        return 0

    if depth >= max_depth:
        return 0

    if maximizing:
        best_score = -1000

        for row in range(len(game_board)):
            for col in range(len(game_board)):
                if game_board[row][col] == 0:
                    game_board[row][col] = 2

                    score = minimax(
                        game_board,
                        depth + 1,
                        max_depth,
                        False
                    )

                    game_board[row][col] = 0
                    best_score = max(best_score, score)

        return best_score

    else:
        best_score = 1000

        for row in range(len(game_board)):
            for col in range(len(game_board)):
                if game_board[row][col] == 0:
                    game_board[row][col] = 1

                    score = minimax(
                        game_board,
                        depth + 1,
                        max_depth,
                        True
                    )

                    game_board[row][col] = 0
                    best_score = min(best_score, score)

        return best_score


def smart_computer_move(game_board, max_depth=5):
    best_score = -1000
    best_move = None

    for row in range(len(game_board)):
        for col in range(len(game_board)):
            if game_board[row][col] == 0:
                game_board[row][col] = 2

                score = minimax(
                    game_board,
                    0,
                    max_depth,
                    False
                )

                game_board[row][col] = 0

                if score > best_score:
                    best_score = score
                    best_move = (row, col)

    if best_move is not None:
        game_board[best_move[0]][best_move[1]] = 2


def smart_computer_game():
    game_board = [[0] * 3 for i in range(3)]

    while winner_verification(game_board) == -1:
        board_game_creation(game_board)
        print("Your turn")

        if player_turn(game_board, 1) == False:
            print("Game cancelled")
            return

        if winner_verification(game_board) != -1:
            break

        print("Computer's turn")
        smart_computer_move(game_board, 5)

    board_game_creation(game_board)

    winner = winner_verification(game_board)

    if winner == 0:
        print("Draw")
    elif winner == 1:
        print("You win!")
    else:
        print("Computer wins!")
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
# Write your solution here
def basic_move(game_board, player):
    other_player = 2 if player == 1 else 1

    # Try to win
    for row in range(len(game_board)):
        for col in range(len(game_board)):
            if game_board[row][col] == 0:
                game_board[row][col] = player

                if winner_verification(game_board) == player:
                    return

                game_board[row][col] = 0

    # Try to block
    for row in range(len(game_board)):
        for col in range(len(game_board)):
            if game_board[row][col] == 0:
                game_board[row][col] = other_player

                if winner_verification(game_board) == other_player:
                    game_board[row][col] = player
                    return

                game_board[row][col] = 0

    # Otherwise first empty spot
    for row in range(len(game_board)):
        for col in range(len(game_board)):
            if game_board[row][col] == 0:
                game_board[row][col] = player
                return


def search_score(game_board, depth, max_depth, smart_player, maximizing):
    result = winner_verification(game_board)
    other_player = 2 if smart_player == 1 else 1

    if result == smart_player:
        return 10 - depth
    elif result == other_player:
        return depth - 10
    elif result == 0:
        return 0

    if depth >= max_depth:
        return 0

    if maximizing:
        best_score = -1000
        player = smart_player

        for row in range(len(game_board)):
            for col in range(len(game_board)):
                if game_board[row][col] == 0:
                    game_board[row][col] = player

                    score = search_score(
                        game_board,
                        depth + 1,
                        max_depth,
                        smart_player,
                        False
                    )

                    game_board[row][col] = 0
                    best_score = max(best_score, score)

        return best_score

    else:
        best_score = 1000
        player = other_player

        for row in range(len(game_board)):
            for col in range(len(game_board)):
                if game_board[row][col] == 0:
                    game_board[row][col] = player

                    score = search_score(
                        game_board,
                        depth + 1,
                        max_depth,
                        smart_player,
                        True
                    )

                    game_board[row][col] = 0
                    best_score = min(best_score, score)

        return best_score


def smarter_move(game_board, player, max_depth):
    best_score = -1000
    best_move = None

    for row in range(len(game_board)):
        for col in range(len(game_board)):
            if game_board[row][col] == 0:
                game_board[row][col] = player

                score = search_score(
                    game_board,
                    0,
                    max_depth,
                    player,
                    False
                )

                game_board[row][col] = 0

                if score > best_score:
                    best_score = score
                    best_move = (row, col)

    if best_move is not None:
        game_board[best_move[0]][best_move[1]] = player


def computer_vs_computer(size, max_depth, games=10):
    smart_wins = 0

    for game in range(games):
        game_board = [[0] * size for i in range(size)]

        # Alternate which player the smart computer controls
        if game % 2 == 0:
            smart_player = 1
            basic_player = 2
        else:
            smart_player = 2
            basic_player = 1

        player = 1

        while winner_verification(game_board) == -1:

            if player == smart_player:
                smarter_move(game_board, player, max_depth)
            else:
                basic_move(game_board, player)

            if player == 1:
                player = 2
            else:
                player = 1

        if winner_verification(game_board) == smart_player:
            smart_wins += 1

    win_rate = (smart_wins / games) * 100

    print(
        size,
        "x",
        size,
        "- Smart player wins:",
        smart_wins,
        "out of",
        games,
        "- Win rate:",
        win_rate,
        "%"
    )
````
