```python
class matrix:
    def __init__(self, n, m= None): #using the m = NOne js in case we do not have a sec arg
        if isinstance(n,list): #is n a list

            row_length = len(n[0]) #checking the value of the col of the first row and putting that into row_len

            for row in n:
                if len(row) != row_length:
                    raise ValueError("The rows length must all be the same")


            self.M = n #storing the matrix n in self.M
        else:
            self.M = [[0 for j in range(m)] for i in range (n)] #the column maker this should make 0 as its char


    def __getitem__(self, key): #key is a holder for wtv we put in []
        if isinstance(key, tuple): #check if key is a tuple
            i, j = key

            if isinstance(i, slice) or isinstance(j, slice): #verification for checking if there is a : in the row or col indexes
                rows = self.M[i] if isinstance(i, slice) else [self.M[i]] #veri

                new_matrix = []

                for row in rows:
                    values = row[j] if isinstance(j, slice) else [row[j]]
                    new_matrix.append(values)

                return matrix(new_matrix)
        
            return self.M[i][j]
        else:
            return self.M[key]

    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self.M[i][j] = value
        else: 
            self.M[key] = value

    def assign(self, other):
        if isinstance(other, matrix):
            values = other.M
        else:
            values = other

        if len(values) != len(self.M):
            raise ValueError("Matrix sizes have to match")
        for i in range(len(self.M)):
            if len(values[i]) != len(self.M[i]):
                raise ValueError("Matrix sizes must match")

        self.M = [row.copy() for row in values]

    def shape(self):
        return (len(self.M), len(self.M[0])) #code for returning the amount of rows we got and how many # that are in the first row

    def transpose(self):
        n, m = self.shape() #look for # of rows and col
        new_matrix = []

        for j in range(m): #go through col of orgi matrix
            new_row = []
            for i in range(n): #go through row of orgi matrix

                new_row.append(self.M[i][j]) #add up values and put into neww row
            new_matrix.append(new_row) #add that new row to matrix

        return matrix(new_matrix)

    def row(self, n):
        return matrix([self.M[n]])

    def column(self, n):
        new_column = []

        for i in range(len(self.M)):
            new_column.append([self.M[i][n]])

        return matrix(new_column)

    def to_list(self):
        return [row.copy() for row in self.M]

    def block(self, n_0, n_1, m_0, m_1): #new varibles for start stop for rows and col
        new_matrix = []

        for i in range (n_0, n_1): #going through slecteted rows
            new_row = []

            for j in range(m_0, m_1):
                new_row.append(self.M[i][j]) #add wanted value

            new_matrix.append(new_row)

        return matrix(new_matrix)
    def scalar_multiple(self, number):
        n, m = self.shape() #pick up for the shape inside is col and row
        answer = matrix(n, m) #creation of 0 matrixs

        for i in range(n):
            for j in range(m):
                answer[i, j] = self.M[i][j] * number

        return answer #this one i didn't read it correctly so i had chat explain why i was getting it wrong and it was bc i did  it in anothher cell


    def add(self, other):
        if self.shape() != other.shape(): #check to see if they have diff dimensions
            raise ValueError("Matrix sizes have to match")

        n, m = self.shape()
        answer = matrix(n,m)

        for i in range(n):
            for j in range(m):
                answer[i, j] = self.M[i][j] + other.M[i][j] #addition of the numbers that are @ the same posion in both of the matrices

        return answer

    def subtract(self, other):
        if self.shape() != other.shape():#check to see if they have diff dimensions
            raise ValueError("Maxtrix sizes need to match")

        n, m = self.shape()
        answer = matrix(n, m)

        for i in range(n):
            for j in range(m):
                answer[i, j] = self.M[i][j] - other.M[i][j]

        return answer

    def multiply(self, other):
        n, m = self.shape()
        p, q = other.shape() #initializing other shape for rows n col

        if m != p:  #check to see if mutilplication is allowed
            raise ValueError("These matrix sizes cannot be multiplied")

        answer = matrix(n, q) #puts  result matrix into answer

        for i in range(n):
            for j in range(q):
                total = 0

                for k in range(m):
                    total = total + self.M[i][k] * other.M[k][j] #mutiplication and addition of the numbers 

                    answer[i, j] = total

        return answer

    def elementwise_multiply(self, other):
        if self.shape() != other.shape():
            raise ValueError("Matrix sizes have to mathch")

        n, m = self.shape()
        answer = matrix(n, m)

        for i in range(n):
            for j in range(m):
                answer[i, j] = self.M[i][j] * other.M[i][j]


        return answer

    def equals(self, other):
        if self.shape() != other.shape():
            return False

        n, m = self.shape()

        for i in range(n):
            for j in range(m):
                if self.M[i][j] != other.M[i][j]:
                    return False

        return True

    def __add__(self, other):
        return self.add(other)

    def __sub__(self, other):
        return self.subtract(other)

    def __mul__(self, other): #scalar mult from left to right
        if isinstance(other, matrix):
            return self.multiply(other)
        else:
            return self.scalar_multiple(other)

    def __eq__(self, other):  #comparision between two matrices
        if not isinstance(other, matrix):
            return False
        return self.equals(other)


    
    
```

```python
def constant(n, m, value): #function for changing  constants varibels to the same num
    M = matrix(n, m)

    for i in range(n):
        for j in range(m):
            M[i, j] = value #inside loop will make the values in the rows n col 5 due to the doubnle  for loop

    return M
```

```python
def zeros(n, m):
    return constant(n, m, 0) #values will return 0 since thats is the value
def ones(n,m):
    return constant(n, m, 1) #values will retrun 1 since thats is the value
```

```python
def eye(n):
    M = zeros(n, n) #code for creating a zero x zero matrix

    for i in range(n):
        M[i, i] = 1

    return M
```
